# Abalone Age Regression

### Beginner-to-Intermediate Machine Learning Assignment

Predict abalone ring count using mixed categorical and numerical features.

This assignment follows an end-to-end machine-learning workflow: focused analysis, preprocessing, data splitting where appropriate, model building, performance measurement, model selection, persistence, and Gradio deployment.


## Dataset Information

**File:** `Abalone.csv`  
**Rows:** 4,177  
**Columns:** 9  
**Target:** `Rings` (age proxy)

**Columns:** Sex, seven physical measurements, and ring count.

**Source:** [UCI Abalone](https://archive.ics.uci.edu/dataset/1/abalone)


## Concepts Covered

- Mixed-type preprocessing
