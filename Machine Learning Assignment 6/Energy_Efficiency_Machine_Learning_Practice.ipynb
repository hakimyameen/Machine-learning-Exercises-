# Building Energy Efficiency Multi-Output Regression

### Beginner-to-Intermediate Machine Learning Assignment

Predict two related building energy targets from design characteristics.

This assignment follows an end-to-end machine-learning workflow: focused analysis, preprocessing, data splitting where appropriate, model building, performance measurement, model selection, persistence, and Gradio deployment.


## Dataset Information

**File:** `EnergyEfficiency.csv`  
**Rows:** 768  
**Columns:** 10  
**Target:** Heating load and cooling load

**Columns:** Eight building-design inputs plus heating and cooling load targets.

**Source:** [UCI Energy Efficiency](https://archive.ics.uci.edu/dataset/242/energy+efficiency)


## Concepts Covered

- Multi-output regression
- per-target metrics
- pipelines
- model tuning
- two-output Gradio UI
