# Wholesale Customer Segmentation

### Beginner-to-Intermediate Machine Learning Assignment

Segment wholesale customers using spending behavior and evaluate unsupervised cluster quality.

This assignment follows an end-to-end machine-learning workflow: focused analysis, preprocessing, data splitting where appropriate, model building, performance measurement, model selection, persistence, and Gradio deployment.


## Dataset Information

**File:** `WholesaleCustomers.csv`  
**Rows:** 440  
**Columns:** 8  
**Target:** No supervised target; customer clusters are discovered

**Columns:** Channel, region, and six annual-spending categories.

**Source:** [UCI Wholesale Customers](https://archive.ics.uci.edu/dataset/292/wholesale+customers)


## Concepts Covered

- Log transformation
- scaling
- K-Means and hierarchical clustering
- silhouette and stability metrics
- cluster deployment
