## Dataset Information

The file **BanknoteAuthentication.csv** contains **1,372 rows** and **5 columns**. Its four input features were produced from wavelet-transformed banknote images. The target contains two classes represented by 0 and 1.

| Column | Role | Description |
|---|---|---|
| `variance` | Input feature | Variance of the wavelet-transformed image |
| `skewness` | Input feature | Skewness of the wavelet-transformed image |
| `curtosis` | Input feature | Kurtosis of the wavelet-transformed image; the source file uses the spelling `curtosis` |
| `entropy` | Input feature | Entropy of the image |
| `class` | Target | Banknote category represented by 0 or 1 |

**Source:** [UCI Machine Learning Repository — Banknote Authentication](https://archive.ics.uci.edu/dataset/267/banknote%2Bauthentication)
