# Breast Cancer Diagnostic Classification

### Beginner-to-Intermediate Machine Learning Assignment

Build an educational classifier for diagnostic categories from numerical measurements. This is not a clinical decision tool.

This assignment follows an end-to-end machine-learning workflow: focused analysis, preprocessing, data splitting where appropriate, model building, performance measurement, model selection, persistence, and Gradio deployment.


## Dataset Information

**File:** `BreastCancerWisconsin.csv`  
**Rows:** 569  
**Columns:** 32  
**Target:** `Diagnosis` (B or M)

**Columns:** ID, 30 numerical cell-nucleus measurements, and diagnosis.

**Source:** [UCI Breast Cancer Wisconsin Diagnostic](https://archive.ics.uci.edu/dataset/17/breast+cancer+wisconsin+diagnostic)


## Concepts Covered

- Numerical classification
- stratified splitting
- scaling and pipelines
- classification metrics
- Gradio table input
