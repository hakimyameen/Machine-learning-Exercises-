# Mushroom Category Classification

### Beginner-to-Intermediate Machine Learning Assignment

Build an educational categorical-classification pipeline. Never use its output to decide whether a real mushroom is safe to eat.

This assignment follows an end-to-end machine-learning workflow: focused analysis, preprocessing, data splitting where appropriate, model building, performance measurement, model selection, persistence, and Gradio deployment.


## Dataset Information

**File:** `Mushroom.csv`  
**Rows:** 8,124  
**Columns:** 23  
**Target:** `poisonous` (e or p)

**Columns:** Twenty-two categorical physical descriptors and an edible/poisonous label.

**Source:** [UCI Mushroom](https://archive.ics.uci.edu/dataset/73/mushroom)
